```python
# Tablas cruzadas (NumPy)
def tabla2x2(a, b):
    t = np.zeros((2,2), dtype=int)
    t[1,1] = np.count_nonzero(a & b)
    t[1,0] = np.count_nonzero(a & ~b)
    t[0,1] = np.count_nonzero(~a & b)
    t[0,0] = np.count_nonzero(~a & ~b)
    return t

tab_py_colab = tabla2x2(prog_py, usa_colab)
tab_py_gh    = tabla2x2(prog_py, cuenta_gh)
tab_colab_gh = tabla2x2(usa_colab, cuenta_gh)
```

```python
import numpy as np
import pandas as pd
```

```python
anios_escolaridad = np.array([0, 3, 6, 9, 12, 16, 18])
ingreso_mensual_estimado_mx = np.array([12621, 15723, 19587, 24401, 30398, 40747, 47176])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(anios_escolaridad, ingreso_mensual_estimado_mx)
plt.title('Años de Escolaridad vs. Ingreso Mensual Estimado')
plt.xlabel('Años de Escolaridad')
plt.ylabel('Ingreso Mensual Estimado (MXN)')
plt.grid(True)
plt.show()
```

Primer intento de modelar los años de escolaridad y el ingreso estimado.   
Utilizar el primer punto y el útlimo para trazar una recta:   

$$ m = \frac{y_{2}-y_{1}} {x_{2}-x_{1}}$$

$$ b = y_{1} - m * x_{1} $$

```python
x1 = anios_escolaridad[0]
y1 = ingreso_mensual_estimado_mx[0]
x2 = anios_escolaridad[-1]
y2 = ingreso_mensual_estimado_mx[-1]

m = (y2 - y1) / (x2 - x1)
b = y1 - m * x1

print(f"La ecuación de la recta es: y = {m:.2f}x + {b:.2f}")
```

```python
linea_ingreso_estimado_new_plot = m * anios_escolaridad + b
linea_ingreso_estimado_new_plot
```

```python

import matplotlib.pyplot as plt

# Calcular los valores y de la recta utilizando m y b
linea_ingreso_estimado_new_plot = m * anios_escolaridad + b

plt.figure(figsize=(10, 6))
plt.scatter(anios_escolaridad, ingreso_mensual_estimado_mx, label='Datos Originales')
plt.plot(anios_escolaridad, linea_ingreso_estimado_new_plot, color='red',
         label=f'Recta de Regresión (y = {m:.2f}x + {b:.2f})')
plt.title('Años de Escolaridad vs. Ingreso Mensual Estimado con Línea de Regresión')
plt.xlabel('Años de Escolaridad')
plt.ylabel('Ingreso Mensual Estimado (MXN)')
plt.grid(True)
plt.legend()
plt.show()
```

Valores predichos $(y_{pred})$: Son los ingresos mensuales que nuestra recta de regresión predice para cada año de escolaridad.   

Errores (residuales): Es la diferencia entre el ingreso mensual real (ingreso_mensual_estimado_mx) y el ingreso mensual predicho (y_pred). Un error positivo significa que la predicción fue menor que el valor real, y un error negativo indica que la predicción fue mayor.   

Errores al cuadrado: Hemos elevado al cuadrado cada uno de los errores. Esto se hace para eliminar los signos negativos (de modo que los errores positivos y negativos no se cancelen entre sí) y para penalizar más los errores grandes.   

Suma de errores cuadrados: Es la suma de todos los errores al cuadrado. Este valor nos da una medida total de la 'distancia' de nuestros puntos a la línea.   

Error Cuadrático Medio (MSE): Es la suma de los errores cuadrados dividida por el número total de puntos de datos. El MSE es una medida promedio de la magnitud de los errores. Un MSE más bajo indica que el modelo se ajusta mejor a los datos.   

En este caso, el Error Cuadrático Medio (MSE) es de aproximadamente 13,196,667.31. Este número nos da una idea del rendimiento general de la recta que trazamos, mostrando la magnitud promedio de los errores al cuadrado entre los valores observados y los valores predichos por la línea.   

Fórmula del MSE (Error Cuadrático Medio) del Inglés Mean Squared Error   

$$ MSE = \frac {\Sigma_{i=1}^{n}(Y_{i}-\hat{Y_{i}})^{2}}{n} $$


Donde:
 $MSE = $ error cuadrático medio.   
 $n = $ número de datos.   
 $ Y_{i} = $ valores observados.   
 $\hat Y_{i} = $ valores predichos.   

```python

# 1. Calcular los valores predichos (y_pred) usando la ecuación de la recta
y_pred = m * anios_escolaridad + b
y_pred
```

```python
# 2. Calcular los errores (residuales): la diferencia entre los valores reales y los predichos
errores = ingreso_mensual_estimado_mx - y_pred
errores
```

Ejercicio 4.  
1. Error Cuadrático Medio (MSE)Mide el promedio de los errores al cuadrado. Al elevar al cuadrado, se eliminan los signos negativos y se penalizan más los errores grandes.$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

2. Error Absoluto Medio (MAE)Es el promedio de las diferencias absolutas entre lo real y lo predicho. Nos da el error promedio en las unidades originales (pesos).$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

3. Raíz del Error Cuadrático Medio (RMSE)Es la raíz cuadrada del MSE. Es muy útil porque devuelve la métrica a las unidades originales de la variable objetivo.$$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$

4. Coeficiente de Determinación ($R^2$)Indica qué proporción de la varianza de los datos es explicada por el modelo. Varía de 0 a 1.$$R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}$$

5. $R^2$ AjustadoAjusta el valor de $R^2$ según el número de variables independientes para evitar el sobreajuste.$$R^2_{adj} = 1 - \left[ \frac{(1 - R^2)(n - 1)}{n - k - 1} \right]$$

6. Error Porcentual Absoluto Medio (MAPE)Representa el error como un porcentaje promedio, facilitando la comprensión de la precisión relativa.$$MAPE = \frac{100\%}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right|$$

```python
#Ejercicio 5

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_real = ingreso_mensual_estimado_mx
y_predicha = y_pred_sklearn
n = len(y_real)
k = 1


mae = mean_absolute_error(y_real, y_predicha)
mse = mean_squared_error(y_real, y_predicha)
rmse = np.sqrt(mse)
r2 = r2_score(y_real, y_predicha)

r2_adj = 1 - ((1 - r2) * (n - 1) / (n - k - 1))


mape = np.mean(np.abs((y_real - y_predicha) / y_real)) * 100

print(f"--- Indicadores de la Regresión Lineal ---")
print(f"MAE:  {mae:.2f}")
print(f"MSE:  {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²:   {r2:.4f}")
print(f"R² Ajustado: {r2_adj:.4f}")
print(f"MAPE: {mape:.2f}%")
```

Preguntas guía para el informe


¿Qué variables son nominales, ordinales, de intervalo y de razón? Da justificación breve.   
Nominales: id_cliente, genero, ciudad, y la derivada alto_ingreso, porque no tienen un valor que pueda ser comparado directamente

Ordinales: segmento y satisfaccion. Porque tienen un orden lógico.   

Intervalo: temperatura_sala. Porque los grados Celsius tienen una escala constante, pero el cero es arbitrario, ya que este no indica la ausencia de calor.

Razón: edad, ingreso_mensual, visitas_ultimo_anio y dias_desde_ultima_visita. Porque poseen un cero absoluto, es decir, indica su ausencia.

¿Qué medidas resumen usaste para cada tipo y por qué?   
Para Nominales: Usamos frecuencias y proporciones (modas). La media es imposible aquí.

Para Ordinales: Se priorizó la mediana y la distribución de frecuencias. Aunque a veces se usa la media en Likert, la mediana es más fiel a la naturaleza jerárquica de los datos.

Para Intervalo/Razón: Usamos media y desviación estándar para ver el promedio, pero añadimos mediana y cuartiles (Q1, Q3) para entender la dispersión y sesgo, especialmente en variables financieras.

¿Qué hallazgos obtuviste al cruzar segmento con ingreso_mensual y con satisfaccion?   
Segmento vs ingreso mensual, y segmento vs satisfacción

¿Hay asimetría notable en ingreso_mensual? ¿Cómo la trataste?   
El ingreso mensual suele presentar una asimetría positiva. Para tratarlo utilizamos el análisis de cuartiles y el boxplot. También se creó la variable dicotómica alto_ingreso (basada en el Q3) para segmentar el comportamiento de los clientes top sin que sus valores extremos afecten los cálculos de tendencia central del resto.

```python
import numpy as np

def make_data(N, err=1.0, rseed=1):
    # randomly sample the data
    rng = np.random.RandomState(rseed)
    X = rng.rand(N, 1) ** 2
    y = 10 - 1. / (X.ravel() + 0.1)
    if err > 0:
        y += err * rng.randn(N)
    return X, y

X, y = make_data(40)
```

```python
import math

# Datos de entrenamiento
X = [(2,4), (4,4), (4,2), (6,2)]
y = ['Rojo', 'Rojo', 'Azul', 'Azul']

# Punto a clasificar
P = (3, 3)
```

```python
def distancia_euclidiana(p1, p2):
    return math.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)
```

Función de distancia euclidiana

```python
distancias = []
for i in range(len(X)):
  d = distancia_euclidiana(P, X[i])
  distancias.append((d, y[i]))

# Ordenar por distancia
distancias.sort(key=lambda x: x[0])

print(distancias)
print("Clase asignada:", distancias[0][1])
```

```python
def visualize_classifier(model, X, y, ax=None, cmap='rainbow'):
    ax = ax or plt.gca()

    # Plot the training points
    ax.scatter(X[:, 0], X[:, 1], c=y, s=30, cmap=cmap,
               clim=(y.min(), y.max()), zorder=3)
    ax.axis('tight')
    ax.axis('off')
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    # fit the estimator
    model.fit(X, y)
    xx, yy = np.meshgrid(np.linspace(*xlim, num=200),
                         np.linspace(*ylim, num=200))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

    # Create a color plot with the results
    n_classes = len(np.unique(y))
    contours = ax.contourf(xx, yy, Z, alpha=0.3,
                           levels=np.arange(n_classes + 1) - 0.5,
                           cmap=cmap, clim=(y.min(), y.max()),
                           zorder=1)

    ax.set(xlim=xlim, ylim=ylim)
```

```python
import pandas as pd

df = pd.read_csv('dataset_liga_mx_jornadas_13_14_15_base.csv')

def obtener_stats(equipo):

    fila = df[(df['equipo_local'] == equipo) | (df['equipo_visita'] == equipo)].tail(1)
    es_local = fila['equipo_local'].values[0] == equipo


    if es_local:
        pts = fila['local_puntos'].values[0]
        racha = fila['local_racha'].values[0]
        pos = fila['local_posicion'].values[0]
    else:
        pts = fila['visita_puntos'].values[0]
        racha = fila['visita_racha'].values[0]
        pos = fila['visita_posicion'].values[0]

    return pts, racha, pos

def crear_df_partido(local, visita):
    pts_l, racha_l, pos_l = obtener_stats(local)
    pts_v, racha_v, pos_v = obtener_stats(visita)

    df_partido = pd.DataFrame({
        'equipo_local': [local], 'equipo_visita': [visita],
        'local_puntos': [pts_l], 'visita_puntos': [pts_v],
        'local_racha': [racha_l], 'visita_racha': [racha_v],
        'local_posicion': [pos_l], 'visita_posicion': [pos_v],
        'diff_puntos': [pts_l - pts_v],
        'diff_racha': [racha_l - racha_v],
        'diff_posicion': [pos_v - pos_l]
    })
    return df_partido


df_queretaro_cruzazul = crear_df_partido('Querétaro', 'Cruz Azul')
df_pumas_juarez       = crear_df_partido('Pumas', 'FC Juárez')
df_monterrey_puebla   = crear_df_partido('Monterrey', 'Puebla')
df_leon_america       = crear_df_partido('León', 'América')
df_atlas_tigres       = crear_df_partido('Atlas', 'Tigres')
df_sanluis_santos     = crear_df_partido('Atlético San Luis', 'Santos')
df_mazatlan_toluca    = crear_df_partido('Mazatlán', 'Toluca')
df_necaxa_chivas      = crear_df_partido('Necaxa', 'Chivas')
df_tijuana_pachuca    = crear_df_partido('Tijuana', 'Pachuca')



features = ['diff_puntos', 'diff_racha', 'diff_posicion']
mapeo = {0: 'Gana visitante', 1: 'Empatan', 2: 'Gana local'}

print(" PREDICCIÓN JORNADA 16 ")
print("1. Querétaro vs Cruz Azul: ", mapeo[rf.predict(df_queretaro_cruzazul[features])[0]])
print("2. Pumas vs FC Juárez:     ", mapeo[rf.predict(df_pumas_juarez[features])[0]])
print("3. Monterrey vs Puebla:    ", mapeo[rf.predict(df_monterrey_puebla[features])[0]])
print("4. León vs América:        ", mapeo[rf.predict(df_leon_america[features])[0]])
print("5. Atlas vs Tigres:        ", mapeo[rf.predict(df_atlas_tigres[features])[0]])
print("6. San Luis vs Santos:     ", mapeo[rf.predict(df_sanluis_santos[features])[0]])
print("7. Mazatlán vs Toluca:     ", mapeo[rf.predict(df_mazatlan_toluca[features])[0]])
print("8. Necaxa vs Chivas:       ", mapeo[rf.predict(df_necaxa_chivas[features])[0]])
print("9. Tijuana vs Pachuca:     ", mapeo[rf.predict(df_tijuana_pachuca[features])[0]])
```

Fitting a support vector machine

```python
from sklearn.svm import SVC # "Support vector classifier"
model = SVC(kernel='linear', C=1E10)
model.fit(X, y)
```

```python
def plot_svc_decision_function(model, ax=None, plot_support=True):
    """Plot the decision function for a 2D SVC"""
    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    # create grid to evaluate model
    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X = np.meshgrid(y, x)
    xy = np.vstack([X.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X.shape)

    # plot decision boundary and margins
    ax.contour(X, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])

    # plot support vectors
    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, facecolors='none');
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
plot_svc_decision_function(model);
```

```python
def plot_svm(N=10, ax=None):
    X, y = make_blobs(n_samples=200, centers=2,
                      random_state=0, cluster_std=0.60)
    X = X[:N]
    y = y[:N]
    model = SVC(kernel='linear', C=1E10)
    model.fit(X, y)

    ax = ax or plt.gca()
    ax.scatter(X[:, 0], X[:, 1], c=y, s=50, cmap='autumn')
    ax.set_xlim(-1, 4)
    ax.set_ylim(-1, 6)
    plot_svc_decision_function(model, ax)

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
fig.subplots_adjust(left=0.0625, right=0.95, wspace=0.1)
for axi, N in zip(ax, [60, 120]):
    plot_svm(N, axi)
    axi.set_title('N = {0}'.format(N))
```

Example: Face Recognition

```python
from sklearn.datasets import fetch_lfw_people
faces = fetch_lfw_people(min_faces_per_person=60)
print(faces.target_names)
print(faces.images.shape)
```

```python
fig, ax = plt.subplots(3, 5)
for i, axi in enumerate(ax.flat):
    axi.imshow(faces.images[i], cmap='bone')
    axi.set(xticks=[], yticks=[],
            xlabel=faces.target_names[faces.target[i]])
```

```python
from sklearn.svm import SVC
from sklearn.decomposition import PCA
from sklearn.pipeline import make_pipeline

pca = PCA(n_components=150, whiten=True, random_state=42)
svc = SVC(kernel='rbf', class_weight='balanced')
model = make_pipeline(pca, svc)
```

```python
from sklearn.model_selection import train_test_split
Xtrain, Xtest, ytrain, ytest = train_test_split(faces.data, faces.target,
                                                random_state=42)
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {'svc__C': [1, 5, 10, 50],
              'svc__gamma': [0.0001, 0.0005, 0.001, 0.005]}
grid = GridSearchCV(model, param_grid)

%time grid.fit(Xtrain, ytrain)
print(grid.best_params_)
```

```python
model = grid.best_estimator_
yfit = model.predict(Xtest)
```

```python
fig, ax = plt.subplots(4, 6)
for i, axi in enumerate(ax.flat):
    axi.imshow(Xtest[i].reshape(62, 47), cmap='bone')
    axi.set(xticks=[], yticks=[])
    axi.set_ylabel(faces.target_names[yfit[i]].split()[-1],
                   color='black' if yfit[i] == ytest[i] else 'red')
fig.suptitle('Predicted Names; Incorrect Labels in Red', size=14);
```

```python
from sklearn.metrics import classification_report
print(classification_report(ytest, yfit,
                            target_names=faces.target_names))
```

```python
from sklearn.metrics import confusion_matrix
mat = confusion_matrix(ytest, yfit)
sns.heatmap(mat.T, square=True, annot=True, fmt='d', cbar=False,
            xticklabels=faces.target_names,
            yticklabels=faces.target_names)
plt.xlabel('true label')
plt.ylabel('predicted label');
```
